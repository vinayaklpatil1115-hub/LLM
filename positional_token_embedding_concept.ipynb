```python
import torch
import torch.nn as nn
```

```python
vocab = {
    'machine' : 0,
    "learning" : 1,
    "python": 2,
    "i": 3,
    "like" : 4,
    "artifical" : 5
}
```

```python
vocab_size = len(vocab)
print(vocab_size)
```

```python
embedding_dim = 7
```

```python
sent = 'i like python'
```

```python
tokens = sent.split()
print(tokens)
```

```python
input_ids = [vocab[word] for word in tokens]
print(input_ids)
```

```python
input_ids = torch.tensor(input_ids)
print(input_ids)
print(type(input_ids))
```

```python
embedding_layer = nn.Embedding(vocab_size, embedding_dim)
```

```python
token_embedding = embedding_layer(input_ids)

print(tokens)
print(input_ids)
print(token_embedding)
```

```python
positional_embedding_layer = nn.Embedding(len(tokens), embedding_dim)
```

```python
position_ids = torch.arange(len(tokens))
print(position_ids)
```

```python
positional_embedding = positional_embedding_layer(position_ids)
print(positional_embedding)
```

```python
final_embedding = token_embedding + positional_embedding
```

```python
print("tokens :",tokens)
print("input IDs :",input_ids)
print("token embedding :" ,token_embedding)
print("position IDs : ",position_ids)
print("Positional embeddings :",positional_embedding)
print("Final Embedding :",final_embedding)
```
