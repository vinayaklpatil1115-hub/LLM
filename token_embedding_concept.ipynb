```python
import torch
import torch.nn as nn
```

```python
vocab = {
    'machine' : 0,
    "learning" : 1,
    "python": 2,
    "i": 3,
    "like" : 4,
    "artifical" : 5
}
```

```python
vocab_size = len(vocab)
print(vocab_size)
```

```python
embedding_dim = 7
```

```python
sent = 'i like python'
```

```python
tokens = sent.split()
print(tokens)
```

```python
input_ids = [vocab[word] for word in tokens]
print(input_ids)
```

```python
print(type(input_ids))
```

```python
input_ids = torch.tensor(input_ids)
print(input_ids)
print(type(input_ids))
```

```python
embedding_layer = nn.Embedding(vocab_size, embedding_dim)
```

```python
token_embedding = embedding_layer(input_ids)

print(tokens)
print(input_ids)
print(token_embedding)
```
