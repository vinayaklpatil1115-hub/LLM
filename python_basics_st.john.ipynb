```python
'''Write a program that checks if someone is eligible for a loan based on the following conditions:

    Age: Must be between 25 and 60 (inclusive).

    Income: Must be greater than or equal to 30,000.

    Credit Score: Must be greater than or equal to 700.

The program should ask for the user’s age, income, and credit score and then print:

    "Eligible for loan" if they meet all criteria

    "Not eligible for loan" otherwise ''
```

```python
age = int(input(" enter your age :"))
income = int(input(" enter your income :"))
credit_score = int(input(" enter your credit score :"))


# nested if-else

if age>=25 and age<=60:
  if income>=30000:
    if credit_score>=700:
      print("eligible for loan")
    else:
      print("not eligible for loan because of bad credit score")
  else:
    print("not eligible for loan beacuse of less income")
else:
  print("not eligible for loan because of bad age")
```
