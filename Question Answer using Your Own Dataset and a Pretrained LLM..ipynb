```python
model = "google/flan-t5-small"
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained(model)
model = AutoModelForSeq2SeqLM.from_pretrained(model)
```

```python
context = '''Python is a high-level programming language.
Python was created by Guido van Rossum.
Python was first released in 1991.
Python is widely used in data science.
Python supports object-oriented programming.
Variables are used to store data.
Python has different data types such as int, float, string and boolean.
Lists are used to store multiple values.
Tuples are ordered and immutable collections.
Dictionaries store data in key-value pairs.
Sets store unique elements.
A function is a reusable block of code.
Functions are defined using the def keyword.
Loops are used to repeat statements.
Python has two main types of loops: for and while.
Conditional statements are used for decision making.
Python supports libraries such as NumPy and Pandas.
NumPy is used for numerical computations.
Pandas is used for data manipulation and analysis.
Machine Learning is one of the applications of Python.'''
```

```python
model.eval()
```

```python
question = input("Ask me question - ")
```

```python
prompt = f'''
        Answer the questions only using the data in context.
        If information is not available in context
        say : Information is not provided

        context : {context}
        question : {question}
        answer :
'''
```

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=512,
    truniction=True
)
```

```python
# generate answer
with torch.no_grad():
  output = model.generate(
      **inputs,
      max_new_tokens = 100,
      num_beams = 4
  )
```

```python
# decode answer
answer = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
print('Question - ', question)
print('Answer - ', answer)
```
