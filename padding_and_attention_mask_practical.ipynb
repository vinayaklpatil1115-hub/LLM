```python
import torch
import torch.nn as nn
from tokenizers import Tokenizer
```

```python
from torch.nn.utils.rnn import pad_sequence
```

```python
tokenizer = Tokenizer.from_file("/content/ai_tokenizer.json")
```

```python
sent = [
    "machine learning is a subset of AI",
    "machine learning",
    "artificial intelligence",
    "neural network layer"
]
```

```python
padding_id = tokenizer.token_to_id("[PAD]")
```

```python
print(padding_id)
```

```python
seq = []

for s in sent:
  encoding = tokenizer.encode(s)
  seq.append(torch.tensor(encoding.ids))

  print(encoding.tokens)
  print(encoding.ids)
```

```python
padding_seq = pad_sequence(seq, batch_first=True, padding_value=padding_id)
```

```python
print(padding_seq)
```

```python
attention_mask = (padding_seq != padding_id).int()
print(attention_mask)
```
