```python
import torch
from torch.nn.utils.rnn import pad_sequence
```

```python
sent_1 = torch.tensor([1,2,3,4,5,6])
sent_2 = torch.tensor([1,2,3])
```

```python
padded = pad_sequence([sent_1, sent_2], batch_first=True, padding_value = 0)
```

```python
print(padded)
```

```python
#attention mask
attention_mask = (padded !=0).int()
print(attention_mask)
```
