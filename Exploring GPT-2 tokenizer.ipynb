```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
words = [
    "happiness",
    "programming",
    "playing",
    "artificialintelligence"
]
```

```python
for word in words:
  tokens = tokenizer.tokenize(word)
  print("Word : ",word)
  print("Tokens : ",tokens)
  print("Token IDs : ",tokenizer.encode(tokens))
  print("----"*10)
```
