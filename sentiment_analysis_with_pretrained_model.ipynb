```python
model = "mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis"
```

```python
from transformers import pipeline
```

```python
classifer = pipeline(
    model = model
    )
```

```python
text = "stock prices collapsed"
```

```python
result = classifer(text)
print(text)
print(result[0]['label'])
print(result[0]["score"])
```
